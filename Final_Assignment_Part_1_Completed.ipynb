This notebook contains the completed code structure for Tasks 1.1–1.9 of the IBM/Coursera automobile-sales visualization assignment.

**Important:** Run **Kernel → Restart & Run All** before uploading to Coursera so that all outputs/graphs are embedded in the notebook.

```python
import numpy as np
import pandas as pd
%matplotlib inline
import matplotlib as mpl
import matplotlib.pyplot as plt
import seaborn as sns
import folium

print('Libraries imported successfully!')
```

```python
# Download the official assignment dataset
URL = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/historical_automobile_sales.csv"

try:
    df = pd.read_csv(URL)
except Exception:
    # JupyterLite-compatible fallback
    from pyodide.http import pyfetch
    import io
    response = await pyfetch(URL)
    data = await response.bytes()
    df = pd.read_csv(io.BytesIO(data))

df['Date'] = pd.to_datetime(df['Date'])
print('Data downloaded and read into a dataframe!')
display(df.head())
```

## TASK 1.1 — Automobile sales fluctuation from year to year

```python
df_line = df.groupby('Year')['Automobile_Sales'].mean()

plt.figure(figsize=(10, 6))
df_line.plot(kind='line', marker='o')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.title('Automobile Sales Over Time')
plt.grid(True)
plt.show()
```

## TASK 1.2 — Sales trends by vehicle type during recession periods

```python
recession_data = df[df['Recession'] == 1]

plt.figure(figsize=(12, 7))
for vehicle_type in recession_data['Vehicle_Type'].unique():
    temp = recession_data[recession_data['Vehicle_Type'] == vehicle_type]
    yearly_sales = temp.groupby('Year')['Automobile_Sales'].mean()
    plt.plot(yearly_sales.index, yearly_sales.values, marker='o', label=vehicle_type)

plt.xlabel('Year')
plt.ylabel('Average Automobile Sales')
plt.title('Automobile Sales Trends by Vehicle Type During Recession Periods')
plt.legend(title='Vehicle Type')
plt.grid(True)
plt.show()
```

## TASK 1.3 — Compare vehicle sales during recession and non-recession periods

```python
plt.figure(figsize=(12, 7))
sns.barplot(data=df, x='Recession', y='Automobile_Sales', hue='Vehicle_Type', errorbar=None)
plt.xlabel('Economic Condition')
plt.ylabel('Average Automobile Sales')
plt.title('Vehicle-Wise Sales During Recession and Non-Recession Periods')
plt.xticks([0, 1], ['Non-Recession', 'Recession'])
plt.legend(title='Vehicle Type')
plt.show()
```

## TASK 1.4 — GDP during recession and non-recession periods

```python
rec_data = df[df['Recession'] == 1]
non_rec_data = df[df['Recession'] == 0]

fig, ax = plt.subplots(1, 2, figsize=(14, 6))

sns.lineplot(data=rec_data, x='Year', y='GDP', ax=ax[0])
ax[0].set_title('GDP During Recession')
ax[0].set_xlabel('Year')
ax[0].set_ylabel('GDP')

sns.lineplot(data=non_rec_data, x='Year', y='GDP', ax=ax[1])
ax[1].set_title('GDP During Non-Recession')
ax[1].set_xlabel('Year')
ax[1].set_ylabel('GDP')

plt.tight_layout()
plt.show()
```

## TASK 1.5 — Bubble plot: seasonality impact on automobile sales

```python
plt.figure(figsize=(12, 7))
sns.scatterplot(
    data=non_rec_data,
    x='Month',
    y='Automobile_Sales',
    size='Seasonality_Weight',
    sizes=(20, 500),
    alpha=0.6,
    legend=True
)
plt.xlabel('Month')
plt.ylabel('Automobile Sales')
plt.title('Seasonality Impact on Automobile Sales')
plt.show()
```

## TASK 1.6 — Scatter plot: vehicle price versus sales during recessions

```python
plt.figure(figsize=(10, 6))
plt.scatter(rec_data['Price'], rec_data['Automobile_Sales'], alpha=0.6)
plt.xlabel('Average Vehicle Price')
plt.ylabel('Automobile Sales')
plt.title('Average Vehicle Price vs Automobile Sales During Recessions')
plt.grid(True)
plt.show()
```

## TASK 1.7 — Advertising expenditure during recession vs non-recession

```python
recession_ad = rec_data['Advertising_Expenditure'].sum()
non_recession_ad = non_rec_data['Advertising_Expenditure'].sum()

plt.figure(figsize=(8, 6))
plt.pie(
    [recession_ad, non_recession_ad],
    labels=['Recession', 'Non-Recession'],
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Advertising Expenditure During Recession and Non-Recession Periods')
plt.show()
```

## TASK 1.8 — Advertising expenditure by vehicle type during recession

```python
ad_by_vehicle = rec_data.groupby('Vehicle_Type')['Advertising_Expenditure'].sum()

plt.figure(figsize=(9, 7))
plt.pie(
    ad_by_vehicle.values,
    labels=ad_by_vehicle.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Share of Each Vehicle Type in Total Advertising Expenditure During Recessions')
plt.show()
```

## TASK 1.9 — Unemployment rate, vehicle type and sales during recession

```python
# The dataset uses the column name unemployment_rate.
# Create the commonly used alias too, so the notebook is easy to follow.
if 'Unemployment_Rate' not in rec_data.columns and 'unemployment_rate' in rec_data.columns:
    rec_data = rec_data.copy()
    rec_data['Unemployment_Rate'] = rec_data['unemployment_rate']

plt.figure(figsize=(12, 7))
sns.lineplot(
    data=rec_data,
    x='Unemployment_Rate',
    y='Automobile_Sales',
    hue='Vehicle_Type'
)
plt.xlabel('Unemployment Rate')
plt.ylabel('Automobile Sales')
plt.title('Effect of Unemployment Rate on Vehicle Type and Sales During Recession')
plt.legend(title='Vehicle Type')
plt.show()
```
